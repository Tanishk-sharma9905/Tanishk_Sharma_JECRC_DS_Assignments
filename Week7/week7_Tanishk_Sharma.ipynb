# Q&A RAG System



```python
from rag_backend import process_document
from rag_backend import answer_question
```

# Document Processing

The uploaded PDF is processed using the backend.
Steps:
1. Read PDF
2. Extract text
3. Split into chunks
4. Generate embeddings
5. Store embeddings inside FAISS

```python
pdf_path = input("Enter PDF Path: ")
vector_db, stats = process_document(pdf_path)
print("Document Processed Successfully")
print("\nStatistics")
for key, value in stats.items():
    print(f"{key}: {value}")
```

# Question Answering

The system retrieves the top 3 relevant chunks and generates a grounded answer.

```python
while True:
    question = input("\nAsk Question (type exit): ")
    if question.lower() == "exit":
        break
    answer, retrieved_chunks = answer_question(vector_db,question)
    print("\nGenerated Answer\n")
    print(answer)
    print("\nRetrieved Context\n")

    for i, chunk in enumerate(retrieved_chunks):
        print(f"\nChunk {i+1}")
        print(chunk)
        print("-"*80)
```
