## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
import re
import logging


logging.basicConfig(level=logging.INFO,format='[%(levelname)s] %(message)s')
logger = logging.getLogger(__name__)
def agent(query):
    """
    Intelligent task-routing agent.

    Routes:
    1. Calculator
    2. Keyword Extraction
    3. General Response

    Returns:
    {
        "type": "...",
        "result": ...
    }
    """

    try:

        if not isinstance(query, str) or not query.strip():
            return {
                "type": "error",
                "result": "Query must be a non-empty string."
            }
        query = query.strip()
        query_lower = query.lower()
        logger.info(f"Received Query: {query}")
        math_pattern = r'^[0-9+\-*/().%\s]+$'

        if "calculate" in query_lower or re.fullmatch(math_pattern, query_lower):
            logger.info("Route Selected -> Calculator")
            if "calculate" in query_lower:
                expression = re.sub(
                    r'^calculate',
                    '',
                    query,
                    flags=re.IGNORECASE
                ).strip()
            else:
                expression = query

            if expression == "":
                return {
                    "type": "error",
                    "result": "No mathematical expression found."
                }
            if not re.search(r'[0-9]', expression):
                return {
                    "type": "error",
                    "result": "Invalid mathematical expression."
                }
            result = calculator(expression)
            return {
                "type": "calculation",
                "result": result
            }


        elif "keywords" in query_lower:
            logger.info("Route Selected -> Keyword Extractor")
            text = re.sub(
                r'^(extract\s+)?keywords(\s+from)?',
                '',
                query,
                flags=re.IGNORECASE
            ).strip()
            if text == "":
                return {
                    "type": "error",
                    "result": "No text provided for keyword extraction."
                }
            result = extract_keywords(text)
            return {
                "type": "keywords",
                "result": result
            }


        else:
            logger.info("Route Selected -> General Response")
            return {
                "type": "general",
                "result": "Sorry, I couldn't identify a supported task. Use 'calculate' for math operations or 'keywords' for keyword extraction."
            }
    except Exception as e:
        logger.exception("Processing Error")
        return {
            "type": "error",
            "result": f"Processing failed: {str(e)}"
        }
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "calculate (12*5)+8",
    "keywords Python programming language is easy to learn",
    "",
    "calculate",
    "Hello there!"
]

for q in queries:
    print("Query :", repr(q))
    print(agent(q))
    print("-" * 50)
````

````python
while True:
    try:
        user_input = input("Enter query (type 'exit' to stop): ")

        if user_input.lower() == "exit":
            print("Goodbye!")
            break

        print("Response:", agent(user_input))

    except KeyboardInterrupt:
        print("\nProgram interrupted. Exiting...")
        break
````

# Single Agent Systems & Agent Pipelines – Quiz

### 1. Explain the concept of a stateful directed graph in agent pipelines. How does it differ from a simple linear pipeline?

A stateful directed graph is a workflow where each step can store and use information from previous steps. It consists of nodes connected by edges, allowing the agent to make decisions, branch into different paths, or repeat tasks when needed. A linear pipeline simply follows a fixed sequence of steps from start to finish without remembering previous states or changing its execution path.

### 2. Describe the role of nodes and edges in an agent workflow. Give an example of each.

Nodes represent the individual tasks performed by an agent, such as analyzing a query, calling a tool, or generating a response. Edges are the connections between nodes that define how information moves through the workflow. For example, a Calculator Tool is a node, while the connection from the Query Analyzer to the Calculator Tool is an edge.

### 3. What is conditional routing in an agent system? Design a simple rule-based routing logic for three different query types.

Conditional routing is the process of sending a query to the most appropriate tool based on its content or intent. A simple routing logic can be:
- If the query contains **"calculate"**, send it to the calculator tool.
- If the query contains **"keywords"**, send it to the keyword extraction tool.
- Otherwise, send it to the general response handler.

### 4. Why are cycles (loops) important in agent pipelines? Provide a use case where a retry loop is necessary.

Loops allow an agent to repeat a task until it succeeds or reaches a stopping condition. They improve reliability by handling temporary failures. For example, if an API request fails because of a network issue, the agent can retry the request a few times before returning an error.

### 5. Explain how a single-agent system can simulate multi-agent behavior internally.

A single-agent system can divide its work into different internal roles. It may first analyze the user's query, decide which tool to use, execute that tool, and then generate the final response. Although these roles appear to be handled by multiple agents, they are all performed by a single system.

### 6. What are JSON schema tools? How do they help in structuring tool inputs and outputs?

JSON schema tools define a standard structure for data exchanged between agents and tools. They specify required fields, data types, and expected formats. This helps validate inputs, produce consistent outputs, reduce errors, and make communication between different components more reliable.

### 7. Compare sequential tool calls and parallel tool calls. When would you prefer one over the other?

Sequential tool calls are executed one after another, with each step depending on the previous one. Parallel tool calls run multiple independent tasks at the same time. Sequential execution is preferred when tasks depend on earlier results, while parallel execution is better for independent tasks because it reduces overall execution time.

### 8. How would you implement error handling in a tool-using agent? Provide at least two strategies.

One strategy is using `try-except` blocks to catch exceptions and return meaningful error messages instead of crashing the program. Another strategy is implementing retry logic for temporary failures such as network issues or API timeouts. Logging errors is also useful because it helps developers identify and fix problems more easily.

### 9. What is trajectory evaluation in agent systems? Why is it important beyond just checking final output?

Trajectory evaluation examines the complete sequence of actions an agent takes while solving a task, including its decisions, tool calls, and intermediate steps. This is important because an agent may produce the correct answer using an inefficient or incorrect process. Evaluating the entire workflow helps identify mistakes, improve decision-making, and optimize overall performance.

### 10. Define task completion rate and cost metrics. How would you measure and optimize them in a real-world system?

Task completion rate is the percentage of tasks that an agent completes successfully. Cost metrics measure the resources used, such as API calls, execution time, token usage, or monetary cost. These metrics can be tracked over many tasks to evaluate performance. They can be optimized by improving routing accuracy, reducing unnecessary tool calls, caching repeated results, and using more efficient algorithms while maintaining a high success rate.
